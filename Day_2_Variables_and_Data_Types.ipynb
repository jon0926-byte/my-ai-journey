```python
# 1. Storing model settings in variables
model_name = "gemini-1.5-pro"  # String: Text data
max_tokens = 500  # Integer: Whole number
temperature = 0.7  # Float: Decimal number
is_active = True  # Boolean: True or False

# 2. Printing the values on the screen
print("--- AI Model Configuration ---")
print(f"Model: {model_name}")
print(f"Max Tokens: {max_tokens}")
print(f"Temperature: {temperature}")
print(f"Is Active: {is_active}")

# 3. Checking the data type of each variable
print("\n--- Variable Data Types ---")
print(type(model_name))
print(type(max_tokens))
print(type(temperature))
print(type(is_active))
```
