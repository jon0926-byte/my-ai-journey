```python
# 1. Define model limit and sample prompt length
max_token_limit = 1000
user_prompt_tokens = 300

print(f"User Prompt Size: {user_prompt_tokens} tokens")
print(f"Max Model Limit: {max_token_limit} tokens")
print("-" * 35)

# 2. Decision logic to check budget
if user_prompt_tokens > max_token_limit:
    print("STATUS: REJECTED")
    print("Reason: Prompt exceeds maximum context window. Truncation required.")
elif user_prompt_tokens >= 700:
    print("STATUS: WARNING")
    print("Reason: Prompt uses over 70% of context window. Near limit.")
else:
    print("STATUS: APPROVED")
    print("Reason: Prompt is well within safe processing limits.")
```
