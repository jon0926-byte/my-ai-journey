```python
# 1. Define model limit and a batch of prompt lengths
max_token_limit = 1000
prompt_batch = [350, 820, 1001, 700, 950]

print("--- STARTING BATCH PROMPT EVALUATION ---")

# 2. Iterate through each prompt in the batch
for index, tokens in enumerate(prompt_batch, start=1):
    print(f"\nPrompt #{index} | Size: {tokens} tokens")

    if tokens > max_token_limit:
        print("STATUS: REJECTED (Exceeds context window limit)")
    elif tokens >= 700:
        print("STATUS: WARNING (Near context limit)")
    else:
        print("STATUS: APPROVED (Safe size)")

print("\n--- BATCH PROCESSING COMPLETE ---")
```
