```python
# 1. Define a custom function to check prompt safety
def evaluate_prompt_length(prompt_text, max_tokens=1000):
    # Estimate token count (rough rule of thumb: ~4 characters per token)
    estimated_tokens = len(prompt_text) // 4

    print(f"Analyzing Prompt ({len(prompt_text)} chars | ~{estimated_tokens} tokens)...")

    if estimated_tokens > max_tokens:
        return "REJECTED: Exceeds context window limit."
    elif estimated_tokens >= 700:
        return "WARNING: Approaching token limit."
    else:
        return "APPROVED: Prompt length is optimal."

# 2. Call the function with different prompts
prompt1 = "Summarize the key points of this customer email."
prompt2 = "Write a comprehensive essay about AI history. " * 50

print("--- PROMPT EVALUATION PIPELINE ---")
result1 = evaluate_prompt_length(prompt1)
print(f"Result 1: {result1}\n")

result2 = evaluate_prompt_length(prompt2)
print(f"Result 2: {result2}")
```
